```python
query = """
SELECT 
    Category,
    "Sub-Category",
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM orders
GROUP BY Category, "Sub-Category"
ORDER BY Total_Revenue DESC;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    Category,
    "Sub-Category",
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(SUM(Profit) * 100.0 / SUM(Revenue), 2) AS Profit_Margin_Percent
FROM orders
GROUP BY Category, "Sub-Category"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    Category,
    "Sub-Category",
    ROUND(AVG(Discount), 3) AS Avg_Discount,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM orders
GROUP BY Category, "Sub-Category"
ORDER BY Avg_Discount DESC;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    "Product Name",
    Sales,
    Quantity,
    Discount,
    Profit,
    Revenue
FROM orders
WHERE "Sub-Category" = 'Supplies'
ORDER BY Profit ASC
LIMIT 15;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    Discount,
    COUNT(*) AS Order_Count,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(AVG(Profit), 2) AS Avg_Profit_Per_Order
FROM orders
WHERE "Sub-Category" = 'Supplies'
GROUP BY Discount
ORDER BY Discount;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    CASE 
        WHEN Discount = 0 THEN '0% (No Discount)'
        WHEN Discount > 0 AND Discount <= 0.1 THEN '1-10%'
        WHEN Discount > 0.1 AND Discount <= 0.2 THEN '11-20%'
        WHEN Discount > 0.2 AND Discount <= 0.3 THEN '21-30%'
        ELSE '30%+'
    END AS Discount_Band,
    COUNT(*) AS Order_Count,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(AVG(Profit), 2) AS Avg_Profit_Per_Order
FROM orders
GROUP BY Discount_Band
ORDER BY MIN(Discount);
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    Region,
    COUNT(*) AS Order_Count,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(SUM(Profit) * 100.0 / SUM(Revenue), 2) AS Profit_Margin_Percent
FROM orders
GROUP BY Region
ORDER BY Total_Profit DESC;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    State,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(SUM(Profit) * 100.0 / SUM(Revenue), 2) AS Profit_Margin_Percent
FROM orders
GROUP BY State
ORDER BY Total_Profit ASC
LIMIT 10;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    State,
    CASE 
        WHEN Discount = 0 THEN '0% (No Discount)'
        WHEN Discount > 0 AND Discount <= 0.1 THEN '1-10%'
        WHEN Discount > 0.1 AND Discount <= 0.2 THEN '11-20%'
        WHEN Discount > 0.2 AND Discount <= 0.3 THEN '21-30%'
        ELSE '30%+'
    END AS Discount_Band,
    COUNT(*) AS Order_Count,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM orders
WHERE State IN ('Texas', 'Ohio')
GROUP BY State, Discount_Band
ORDER BY State, MIN(Discount);
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    strftime('%Y', "Order Date") AS Year,
    strftime('%m', "Order Date") AS Month,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM orders
GROUP BY Year, Month
ORDER BY Year, Month;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    Segment,
    COUNT(*) AS Order_Count,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(SUM(Profit) * 100.0 / SUM(Revenue), 2) AS Profit_Margin_Percent,
    ROUND(AVG(Discount), 3) AS Avg_Discount
FROM orders
GROUP BY Segment
ORDER BY Total_Profit DESC;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
    "Customer Name",
    COUNT(*) AS Order_Count,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM orders
GROUP BY "Customer Name"
ORDER BY Total_Profit DESC
LIMIT 10;
"""
pd.read_sql(query, conn)
```

```python
query = """
WITH customer_profit AS (
    SELECT 
        "Customer Name",
        SUM(Profit) AS Total_Profit
    FROM orders
    GROUP BY "Customer Name"
    ORDER BY Total_Profit DESC
)
SELECT 
    "Customer Name",
    ROUND(Total_Profit, 2) AS Total_Profit,
    ROUND(SUM(Total_Profit) OVER (ORDER BY Total_Profit DESC) * 100.0 / 
        (SELECT SUM(Total_Profit) FROM customer_profit), 2) AS Cumulative_Profit_Percent
FROM customer_profit
ORDER BY Total_Profit DESC
LIMIT 20;
"""
pd.read_sql(query, conn)
```

```python
query = """
WITH customer_profit AS (
    SELECT 
        "Customer Name",
        SUM(Profit) AS Total_Profit
    FROM orders
    GROUP BY "Customer Name"
),
ranked AS (
    SELECT 
        "Customer Name",
        Total_Profit,
        SUM(Total_Profit) OVER (ORDER BY Total_Profit DESC) * 100.0 / 
            (SELECT SUM(Total_Profit) FROM customer_profit) AS Cumulative_Percent,
        ROW_NUMBER() OVER (ORDER BY Total_Profit DESC) AS Rank
    FROM customer_profit
)
SELECT MIN(Rank) AS Customers_Needed_For_80_Percent
FROM ranked
WHERE Cumulative_Percent >= 80;
"""
pd.read_sql(query, conn)

# total unique customers bhi nikal lete hain comparison ke liye
total_customers_query = 'SELECT COUNT(DISTINCT "Customer Name") AS Total_Customers FROM orders;'
pd.read_sql(total_customers_query, conn)
```
